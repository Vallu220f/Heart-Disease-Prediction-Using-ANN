## 4. Data preprocessing

```python
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
```

```python
numerical_cols = ['BMI', 'MentHlth', 'PhysHlth']

binary_cols = ['HighBP', 'HighChol', 'CholCheck', 'Smoker', 'Stroke', 'PhysActivity',

               'Fruits', 'Veggies', 'HvyAlcoholConsump', 'AnyHealthcare', 'NoDocbcCost',

               'DiffWalk', 'Sex']





discrete_cols = ['Diabetes', 'GenHlth', 'Age', 'Education', 'Income']
```

```python
num_pipe = Pipeline([('scaling', StandardScaler())])
preprocessor = ColumnTransformer([

    ('scaling', num_pipe, numerical_cols),

    ('discrete', 'passthrough', discrete_cols),

    ('binary', 'passthrough', binary_cols)

])

preprocessor
```

## 5. Train / Validation / Test split

```python
from sklearn.model_selection import train_test_split



x = df.drop(columns='HeartDiseaseorAttack')

y = df['HeartDiseaseorAttack']



xtrain, xtemp, ytrain, ytemp = train_test_split(

    x, y, test_size=0.30, stratify=y, random_state=23

)

xval, xtest, yval, ytest = train_test_split(

    xtemp, ytemp, test_size=0.50, stratify=ytemp, random_state=23

)

xtrain.shape, xval.shape, xtest.shape
```

```python
xtrain_prep = preprocessor.fit_transform(xtrain)

xval_prep = preprocessor.transform(xval)

xtest_prep = preprocessor.transform(xtest)
```

## 6. Handle class imbalance

```python
from sklearn.utils.class_weight import compute_class_weight



classes = np.unique(ytrain)

class_weights = compute_class_weight(class_weight='balanced', classes=classes, y=ytrain)

class_weights = dict(zip(classes, class_weights))

class_weights
```

## 7. Baseline ANN

```python
from keras.models import Sequential

from keras.layers import Input, Dense

from keras.initializers import he_normal

from keras.optimizers import Adam

from keras.metrics import Recall, AUC
```

```python
model = Sequential()


model.add(Input(shape=(xtrain_prep.shape[1],)))


model.add(Dense(64, activation='relu', kernel_initializer=he_normal()))

model.add(Dense(32, activation='relu', kernel_initializer=he_normal()))

model.add(Dense(32, activation='relu', kernel_initializer=he_normal()))

model.add(Dense(16, activation='relu', kernel_initializer=he_normal()))

model.add(Dense(1, activation='sigmoid'))



model.summary()
```

```python

model.compile(

    optimizer='Adam',

    loss='binary_crossentropy',

    metrics=['accuracy', Recall(name='recall'), AUC(name='auc')]

)
```

```python
history = model.fit(

    xtrain_prep, ytrain,

    epochs=15,

    batch_size=32,

    validation_data=(xval_prep, yval),

    class_weight=class_weights

)
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
sns.lineplot(x=history.epoch, y=history.history['accuracy'], label='training accuracy')

sns.lineplot(x=history.epoch, y=history.history['val_accuracy'], label='validation accuracy')

plt.show()
```

```python
sns.lineplot(x=history.epoch, y=history.history['recall'], label='training recall')

sns.lineplot(x=history.epoch, y=history.history['val_recall'], label='validation recall')

plt.show()
```

```python
sns.lineplot(x=history.epoch, y=history.history['loss'], label='training loss')

sns.lineplot(x=history.epoch, y=history.history['val_loss'], label='validation loss')

plt.show()
```

```python
from sklearn.metrics import classification_report, confusion_matrix, roc_auc_score



y_prob = model.predict(xtest_prep)

y_pred = (y_prob > 0.5).astype(int)



print(classification_report(ytest, y_pred))

print(confusion_matrix(ytest, y_pred))

print('ROC-AUC:', roc_auc_score(ytest, y_prob))
```

## 9. Hyperparameter tuning with Optuna

```python
import optuna

from keras.optimizers import Adam, RMSprop

from keras.layers import Dropout, BatchNormalization

from keras.regularizers import L1L2

from keras.callbacks import EarlyStopping
```

```python
def objective(trial):

    n_layers = trial.suggest_int('n_layers', 1, 10)

    epochs = trial.suggest_int('epochs', 35, 150)

    lr_rate = trial.suggest_float('lr_rate', 1e-3, 1e-2, log=True)

    batch_size = trial.suggest_int('batch_size', 64, 128)

    optimizer_name = trial.suggest_categorical('optimizers', ['Adam', 'RMSprop'])



    model = Sequential()

    model.add(Input(shape=(xtrain_prep.shape[1],)))



    for i in range(n_layers):

        n_units = trial.suggest_int(f'units{i}', 4, 200)

        dropout = trial.suggest_float(f'dropout{i}', 0.1, 0.5)

        strength = trial.suggest_float(f'regularization{i}', 1e-3, 1e-2, log=True)



        model.add(Dense(units=n_units, activation='relu', kernel_initializer=he_normal(),

                        kernel_regularizer=L1L2(l1=strength, l2=strength)))

        model.add(Dropout(rate=dropout))

        model.add(BatchNormalization())



    model.add(Dense(1, activation='sigmoid'))



    early_stopping = EarlyStopping(monitor='val_auc', mode='max', min_delta=0.001, patience=3,

                                    restore_best_weights=True)

    optimizer = Adam(learning_rate=lr_rate) if optimizer_name == 'Adam' else RMSprop(learning_rate=lr_rate)



    model.compile(optimizer=optimizer, loss='binary_crossentropy',

                  metrics=['accuracy', Recall(name='recall'), AUC(name='auc')])





    trial_history = model.fit(

        xtrain_prep, ytrain,

        epochs=epochs,

        batch_size=batch_size,

        validation_data=(xval_prep, yval),

        class_weight=class_weights,

        callbacks=[early_stopping],

        verbose=0

    )





    return max(trial_history.history['val_auc'])
```

```python
study = optuna.create_study(direction='maximize')

study.optimize(objective, n_trials=5)
```

```python
study.best_params
```

## training the model

```python
best_params = study.best_params



n_layers = best_params['n_layers']

epochs = best_params['epochs']

lr_rate = best_params['lr_rate']

batch_size = best_params['batch_size']

optimizer_name = best_params['optimizers']



final_model = Sequential()

final_model.add(Input(shape=(xtrain_prep.shape[1],)))



for i in range(n_layers):

    units = best_params[f'units{i}']

    dropout = best_params[f'dropout{i}']

    strength = best_params[f'regularization{i}']



    final_model.add(Dense(units=units, activation='relu', kernel_initializer=he_normal(),

                          kernel_regularizer=L1L2(l1=strength, l2=strength)))

    final_model.add(Dropout(dropout))

    final_model.add(BatchNormalization())



final_model.add(Dense(1, activation='sigmoid'))



final_optimizer = Adam(learning_rate=lr_rate) if optimizer_name == 'Adam' else RMSprop(learning_rate=lr_rate)



early_stopping = EarlyStopping(monitor='val_auc', mode='max', min_delta=0.001, patience=10,

                                restore_best_weights=True)



final_model.compile(optimizer=final_optimizer, loss='binary_crossentropy',

                    metrics=['accuracy', Recall(name='recall'), AUC(name='auc')])
```

```python
final_history = final_model.fit(

    xtrain_prep, ytrain,

    epochs=epochs,

    batch_size=batch_size,

    validation_data=(xval_prep, yval),

    class_weight=class_weights,

    callbacks=[early_stopping]

)
```

```python
final_model.summary()
```

```python
sns.lineplot(x=final_history.epoch, y=final_history.history['accuracy'], label='training accuracy')

sns.lineplot(x=final_history.epoch, y=final_history.history['val_accuracy'], label='validation accuracy')

plt.show()
```

```python
sns.lineplot(x=final_history.epoch, y=final_history.history['loss'], label='training loss')

sns.lineplot(x=final_history.epoch, y=final_history.history['val_loss'], label='validation loss')

plt.show()
```

## 11. Final 

```python
y_prob_final = final_model.predict(xtest_prep)

y_pred_final = (y_prob_final > 0.5).astype(int)

print(classification_report(ytest, y_pred_final))

print(confusion_matrix(ytest, y_pred_final))

print('ROC-AUC:', roc_auc_score(ytest, y_prob_final))
```
